### Scenario
Wanjiku runs a fresh-produce stall at Gikomba Market and has recorded daily sales and stock information for two weeks. This analysis uses core Python to evaluate sales performance, compare the two weeks, classify daily performance, compare weekdays and weekends, identify products that need restocking, and calculate gross margins. The goal is to provide practical, data-backed recommendations that can help Wanjiku improve sales and manage inventory.

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]   # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

## Task 1: Overall Sales Performance

I will calculate the total, average, highest, and lowest daily sales for the 14-day period. I will also use the position of the highest and lowest sales to identify the corresponding day name and week number.

```python
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

highest_index = daily_sales.index(highest_sales)
lowest_index = daily_sales.index(lowest_sales)

highest_day = days[highest_index]
lowest_day = days[lowest_index]

highest_week = 1 if highest_index < 7 else 2
lowest_week = 1 if lowest_index < 7 else 2

print("Task 1: Overall Sales Performance")
print(f"Total sales: KES {total_sales:,}")
print(f"Average daily sales: KES {average_sales:,.2f}")
print(f"Highest daily sales: KES {highest_sales:,}")
print(f"Highest sales day: {highest_day}, Week {highest_week}")
print(f"Lowest daily sales: KES {lowest_sales:,}")
print(f"Lowest sales day: {lowest_day}, Week {lowest_week}")
```

## Task 2: Weekly Sales Comparison

I will create a week_totals() function that uses slicing to separate the 14 days into Week 1 and Week 2. I will then calculate the percentage growth from Week 1 to Week 2.

```python
def week_totals(sales):
    """Return the total sales for Week 1 and Week 2."""
    week_1 = sum(sales[:7])
    week_2 = sum(sales[7:])
    return week_1, week_2


week_1_total, week_2_total = week_totals(daily_sales)

week_growth = ((week_2_total - week_1_total) / week_1_total) * 100

print("Task 2: Weekly Sales Comparison")
print(f"Week 1 total: KES {week_1_total:,}")
print(f"Week 2 total: KES {week_2_total:,}")
print(f"Week-on-week growth: {week_growth:.2f}%")
```

## Task 3: Daily Performance Classification

I will classify each day as Strong, Normal, or Weak based on its sales compared with the overall average. A day is Strong when sales are more than 20% above the average, Weak when sales are more than 20% below the average, and Normal otherwise. I will also count the number of days in each category using a dictionary.

```python
def classify_day(amount, average):
    """Classify a sales amount as Strong, Normal, or Weak."""
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"


category_counts = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

daily_classifications = []

for day, amount in zip(days, daily_sales):
    category = classify_day(amount, average_sales)
    category_counts[category] += 1
    daily_classifications.append((day, amount, category))

print("Task 3: Daily Performance Classification")

for day, amount, category in daily_classifications:
    print(f"{day}: KES {amount:,} - {category}")

print("\nCategory counts:")
for category, count in category_counts.items():
    print(f"{category}: {count} days")
```

## Task 4: Weekday versus Weekend Performance

I will separate the sales into weekdays (Monday to Friday) and weekends (Saturday and Sunday). I will calculate the average for each group and then determine how much higher or lower weekend average sales are compared with weekday average sales.

```python
weekday_sales = []
weekend_sales = []

for day, amount in zip(days, daily_sales):
    if day in ["Mon", "Tue", "Wed", "Thu", "Fri"]:
        weekday_sales.append(amount)
    else:
        weekend_sales.append(amount)

weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

weekend_difference = (
    (weekend_average - weekday_average) / weekday_average
) * 100

print("Task 4: Weekday versus Weekend Performance")
print(f"Weekday average sales: KES {weekday_average:,.2f}")
print(f"Weekend average sales: KES {weekend_average:,.2f}")
print(f"Weekend difference from weekday average: {weekend_difference:.2f}%")
```

## Task 5: Inventory Reorder and Gross Margin Analysis

I will create a function that checks the stock dictionary and returns products whose current quantity is at or below their reorder level. I will also calculate the gross margin percentage for every product using the formula (price - cost) / price × 100.

```python
def products_to_reorder(stock_data):
    """Return products whose quantity is at or below the reorder level."""
    reorder_products = []

    for product, details in stock_data.items():
        if details["qty"] <= details["reorder"]:
            reorder_products.append(product)

    return reorder_products


reorder_list = products_to_reorder(stock)

print("Task 5: Inventory and Gross Margin Analysis")

print("Products to reorder:")
for product in reorder_list:
    print(f"- {product}")

print("\nGross margins:")

for product, details in stock.items():
    gross_margin = (
        (details["price"] - details["cost"]) / details["price"]
    ) * 100

    print(f"{product}: {gross_margin:.2f}%")
```

## Task 6: Stall Performance Report

I will combine the main results from the analysis into a formatted business report. F-strings will be used to present KES values with thousands separators and percentages with appropriate decimal places.

```python
print("=" * 50)
print("          STALL PERFORMANCE REPORT")
print("=" * 50)

print(f"Total 14-day sales:       KES {total_sales:,}")
print(f"Average daily sales:      KES {average_sales:,.2f}")
print(f"Week 1 sales:             KES {week_1_total:,}")
print(f"Week 2 sales:             KES {week_2_total:,}")
print(f"Week-on-week growth:      {week_growth:.2f}%")
print(f"Best sales day:            {highest_day}, Week {highest_week}")
print(f"Best daily sales:         KES {highest_sales:,}")
print(f"Worst sales day:           {lowest_day}, Week {lowest_week}")
print(f"Worst daily sales:        KES {lowest_sales:,}")
print(f"Weekday average:          KES {weekday_average:,.2f}")
print(f"Weekend average:          KES {weekend_average:,.2f}")
print(f"Weekend advantage:        {weekend_difference:.2f}%")
print(f"Strong days:              {category_counts['Strong']}")
print(f"Normal days:              {category_counts['Normal']}")
print(f"Weak days:                {category_counts['Weak']}")
print(f"Products to reorder:      {', '.join(reorder_list)}")

print("\nGross Margins:")
for product, details in stock.items():
    gross_margin = (
        (details["price"] - details["cost"]) / details["price"]
    ) * 100
    print(f"{product:<12} {gross_margin:>6.2f}%")

print("=" * 50)
```
